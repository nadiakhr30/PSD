# Business Understanding

## 1. Latar Belakang
Dalam bidang florikultura dan penelitian biologi, identifikasi spesies bunga merupakan langkah penting untuk mengoptimalkan budidaya, distribusi, dan riset. Dataset Iris mencakup tiga spesies dominan: Setosa, Versicolor, dan Virginica, yang dibedakan berdasarkan ukuran sepal dan petal. Data ini bersifat numerik, namun variasi antar individu membuat klasifikasi manual sering tidak konsisten, sehingga analisis berbasis data menjadi solusi untuk memahami karakter morfologi bunga secara akurat.

## 2. Business Problem
Bagaimana data pengukuran (panjang dan lebar sepal serta petal) dapat digunakan untuk:
- Memeriksa missing values, memastikan tidak ada data yang hilang.
- Mengidentifikasi outlier yang dapat memengaruhi analisis statistik.
- Menyeimbangkan distribusi data antar spesies bila diperlukan untuk model prediktif.
- Memberikan deskripsi numerik dan visual dari setiap atribut, sehingga menjadi dasar analisis klasifikasi lebih lanjut.

## 3. Tujuan Analisis
- Menentukan tipe data tiap atribut dan struktur dataset secara keseluruhan.
- Menghitung statistik dasar (min, max, rata-rata, penyebaran) dan mendeteksi outlier.
- Memeriksa apakah dataset memiliki missing values atau perlu penyeimbangan data antar spesies.
- Menampilkan distribusi jumlah tiap spesies menggunakan visualisasi yang sesuai.
- Memberikan insight awal yang mendukung pembangunan model klasifikasi pada tahap berikutnya.

## 4. Manfaat
- Memberikan pemahaman awal untuk analisis prediktif dalam membedakan spesies bunga Iris.
- Mempercepat proses identifikasi bunga tanpa harus mengandalkan observasi manual.
- Memberikan nilai tambah bagi peneliti, petani, dan industri florikultura dalam hal efisiensi dan akurasi.
- Mendukung pengambilan keputusan berbasis data di bidang pertanian, bioteknologi, hingga konservasi tanaman.
